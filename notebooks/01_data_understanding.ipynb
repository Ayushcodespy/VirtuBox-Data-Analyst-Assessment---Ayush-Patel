```python
df = pd.concat(
    [df_2009, df_2010],
    ignore_index=True
)

print("Combined Shape:", df.shape)
```

```python
missing_summary = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (
        df.isnull().mean() * 100
    ).round(2)
})

missing_summary = missing_summary.sort_values(
    by="Missing Percentage",
    ascending=False
)

missing_summary
```

```python
missing_description["StockCode"].value_counts().head(20)
```

```python
description_mapping = (
    df.dropna(subset=["Description"])
      .groupby("StockCode")["Description"]
      .nunique()
)

print(
    "StockCodes with multiple descriptions:",
    (description_mapping > 1).sum()
)
```

```python
available_description_codes = set(
    df.loc[
        df["Description"].notna(),
        "StockCode"
    ]
)

missing_description_codes = set(
    missing_description["StockCode"]
)

recoverable_codes = (
    missing_description_codes
    & available_description_codes
)

print(
    "Unique StockCodes with missing descriptions:",
    len(missing_description_codes)
)

print(
    "StockCodes with descriptions available elsewhere:",
    len(recoverable_codes)
)
```

```python
# Count unique non-missing descriptions for each StockCode
description_mapping = (
    df.dropna(subset=["Description"])
      .groupby("StockCode")["Description"]
      .nunique()
)

# Keep only StockCodes having exactly one unique description
consistent_stockcodes = description_mapping[
    description_mapping == 1
].index

print(
    "StockCodes with a consistent description:",
    len(consistent_stockcodes)
)
```

```python
consistent_description_map = (
    df[df["StockCode"].isin(consistent_stockcodes)]
    .dropna(subset=["Description"])
    .drop_duplicates(subset=["StockCode"])
    .set_index("StockCode")["Description"]
)

consistent_description_map.head(10)
```

```python
recoverable_mask = (
    df["Description"].isnull()
    & df["StockCode"].isin(consistent_description_map.index)
)

print(
    "Missing descriptions that can be recovered:",
    recoverable_mask.sum()
)

print(
    "Missing descriptions that cannot be safely recovered:",
    df["Description"].isnull().sum() - recoverable_mask.sum()
)
```

```python
df_clean = df.copy()

df_clean["Description"] = df_clean["Description"].fillna(
    df_clean["StockCode"].map(consistent_description_map)
)
```

```python
print("Before:", df["Description"].isnull().sum())

print("After:", df_clean["Description"].isnull().sum())
```
