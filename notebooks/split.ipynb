```python
import pandas as pd
import os

# Cleaned dataset load karo
df_clean = pd.read_csv("../data/processed/online_retail_cleaned.csv")

# Output folder
output_dir = "../data/splits/split_data"
os.makedirs(output_dir, exist_ok=True)

# 200,000 rows per file
chunk_size = 200_000

for i in range(0, len(df_clean), chunk_size):
    chunk = df_clean.iloc[i:i + chunk_size]

    file_number = (i // chunk_size) + 1
    file_path = os.path.join(
        output_dir,
        f"online_retail_cleaned_part_{file_number}.csv"
    )

    chunk.to_csv(file_path, index=False)

    print(f"Saved: {file_path}")
```
